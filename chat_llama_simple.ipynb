```python
from langchain_ollama import OllamaLLM as Ollama
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt=ChatPromptTemplate.from_messages([("system",
"You are helpful assistant.please respond like soft person to user queries"),
                                         ("user","Question:{question}")])
```

```python
model=Ollama(model="llama3.2")
```

```python
output_parser = StrOutputParser()
```

```python
chain=prompt|model|output_parser
```

```python
user_question="What is the capital city of india?"
```

```python
try:
    response=chain.invoke({"question":user_question})
    print(response)
except Exception as e:
    print(f"An Error occured:{e}")
    print("Make sure that youhave ollama installed and running and that the model is downloaded and correct!")
    print("RUn ollama pull llama3.2 to download the model ifyou hvnt done already! ")
```

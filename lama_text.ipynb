```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=10)
```

```python
print(type(text_splitter))
```

```python
text_document_splits=text_splitter.split_documents(text_document)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
embeddings=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
db=FAISS.from_documents(text_document_splits, embeddings)
```

```python
query="when did economic liberalization reforms begin in India?"
```

```python
docs=db.similarity_search(query)
```
